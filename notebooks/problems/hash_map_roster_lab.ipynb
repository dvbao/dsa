re-hash every existing entry into more buckets, keeping lookups fast.

## Part 2: Linear Probing

`HashMap` above resolves collisions with **separate chaining**: each
bucket is a list that can hold more than one entry. There's a very
different approach called **open addressing**: every entry lives directly
inside the table itself, one entry per slot. When the slot a key hashes to
is already taken, you *probe* forward to the next slot instead of
chaining.

Implement `LinearProbingMap` below using **linear probing**: if slot
`_hash(key)` is occupied, try `_hash(key) + 1`, then `+2`, wrapping around
with `% capacity`, until you find an open slot.

**Why deletion needs a "tombstone."** Say key `B` wanted slot 4 but it was
taken, so `B` ended up in slot 5. Later, whatever was in slot 4 gets
deleted and that slot is reset straight back to empty. Now searching for
`B` starts at slot 4, sees "empty," and incorrectly concludes `B` isn't
there, even though it's sitting right next door in slot 5. The fix:
deleting a slot sets it to a special `AVAILABLE` marker (provided below)
instead of `None`. A search keeps probing through `AVAILABLE` slots
("something used to be here, keep looking") and only stops at a truly
empty (`None`) slot.

Note `LOAD_FACTOR_THRESHOLD` is lower here (0.5, vs. 0.75 for chaining):
open addressing gets slow much faster as the table fills up, since a full
table has nowhere left to probe to.

This is a separate exercise from `HashMap` above: it doesn't touch or
depend on it. The roster app below has a **Backend** switch so you can
test and compare both.

````python
# _roster holds the live map instance, and _backend tracks which class it
# should be. Both are built/switched lazily (on first use, not when this
# cell runs) so the UI below always renders even if neither class is
# implemented yet.
_roster = {"instance": None}
_backend = {"name": "chaining"}


def current_class():
    return HashMap if _backend["name"] == "chaining" else LinearProbingMap


def get_roster():
    if _roster["instance"] is None:
        _roster["instance"] = current_class()()
    return _roster["instance"]


# Visual shell: pure decoration, no logic lives here.
BORDER = "1px solid #d0d7de"

TITLE_BAR = widgets.HTML("""
<div style="
    background: linear-gradient(135deg, #3B6EA5, #24456B);
    color: #ffffff;
    padding: 16px 20px;
    border-radius: 8px 8px 0 0;
    font-family: -apple-system, Segoe UI, Helvetica, Arial, sans-serif;
">
  <div style="font-size: 20px; font-weight: 600;">Course Roster Manager</div>
  <div style="font-size: 13px; opacity: 0.85; margin-top: 2px;">
    Registrar backend, powered by the map class you write above
  </div>
</div>
""")

FOOTER = widgets.HTML("""
<div style="font-size: 11px; color: #8a8a8a; margin-top: 10px; text-align: right;
            font-family: -apple-system, Segoe UI, Helvetica, Arial, sans-serif;">
  Buttons always run. An unfinished method just says so instead of crashing.
</div>
""")


def section_header(text):
    return widgets.HTML(f"""
    <div style="
        border-left: 4px solid #3B6EA5;
        padding: 3px 10px;
        margin: 14px 0 6px 0;
        font-weight: 600;
        font-family: -apple-system, Segoe UI, Helvetica, Arial, sans-serif;
        color: #24456B;
    ">{text}</div>
    """)


# Live widgets.
backend_toggle = widgets.ToggleButtons(
    options=[("Separate Chaining", "chaining"), ("Linear Probing", "probing")],
    value="chaining",
    description="Backend:",
)

id_box = widgets.IntText(description="Student ID:", value=1001,
                          layout=widgets.Layout(width="260px"))
name_box = widgets.Text(description="Name:", placeholder="Alice Nguyen",
                         layout=widgets.Layout(width="320px"))
gpa_box = widgets.FloatText(description="GPA:", value=4.0,
                             layout=widgets.Layout(width="220px"))

enroll_btn = widgets.Button(description="Enroll / Update", button_style="success", icon="user-plus")
lookup_btn = widgets.Button(description="Look Up", button_style="info", icon="search")
drop_btn = widgets.Button(description="Drop Student", button_style="danger", icon="user-times")
list_btn = widgets.Button(description="List Roster", icon="list")
load_btn = widgets.Button(description="Load Sample Data", icon="database")
reset_btn = widgets.Button(description="Reset Roster", icon="refresh")
clear_log_btn = widgets.Button(description="Reset Activity Log", icon="eraser")

# No fixed height / scrollbar here on purpose: everything stays fully
# visible (no hidden overflow) so screenshots and browser printing always
# show the complete log and the complete bucket/slot list, not just
# whatever fit in a small scrollable box.
out = widgets.Output(layout=widgets.Layout(
    border_top=BORDER, border_right=BORDER, border_bottom=BORDER, border_left=BORDER,
    padding="8px",
))
internals_out = widgets.Output(layout=widgets.Layout(
    border_top=BORDER, border_right=BORDER, border_bottom=BORDER, border_left=BORDER,
    padding="8px",
))


def log(msg):
    with out:
        print(msg)


def refresh_internals():
    internals_out.clear_output()
    with internals_out:
        inst = _roster["instance"]
        if inst is None:
            print("(roster not created yet, click any button to try)")
            return
        try:
            table = inst._table
            size = inst._size
            capacity = len(table)
            load_factor = size / capacity if capacity else 0.0
            print(f"backend = {_backend['name']}   capacity = {capacity}   "
                  f"size = {size}   load factor = {load_factor:.2f}")

            if _backend["name"] == "chaining":
                for i, bucket in enumerate(table):
                    if bucket:
                        keys = ", ".join(repr(k) for k, _ in bucket)
                        tag = "  <-- collision!" if len(bucket) > 1 else ""
                        print(f"  [{i:>2}] {keys}{tag}")
                    else:
                        print(f"  [{i:>2}] (empty)")
            else:
                for i, slot in enumerate(table):
                    if slot is None:
                        print(f"  [{i:>2}] (empty)")
                    elif slot is AVAILABLE:
                        print(f"  [{i:>2}] (deleted)")
                    else:
                        k, _ = slot
                        tag = ""
                        try:
                            ideal = inst._hash(k)
                            if ideal != i:
                                tag = f"  <-- probed +{(i - ideal) % capacity} past slot {ideal}"
                        except Exception:
                            pass
                        print(f"  [{i:>2}] {k!r}{tag}")
        except AttributeError:
            print("(internal state not available yet, make sure __init__ "
                  "sets self._table and self._size)")
        except Exception as e:
            print(f"(internals unavailable: {type(e).__name__}: {e})")


def guarded(handler):
    """Run a handler, turning unfinished methods / bad input into a
    friendly log message instead of a crash, then refresh the internals
    panel no matter what happened."""
    def wrapped(_btn):
        try:
            handler()
        except NotImplementedError:
            log(f"Not implemented yet: finish {current_class().__name__} above, "
                "re-run that cell, then try again.")
        except KeyError:
            log(f"No student with ID {id_box.value}.")
        except Exception as e:
            log(f"Error: {type(e).__name__}: {e}")
        finally:
            refresh_internals()
    return wrapped


def on_backend_change(change):
    if change["name"] == "value":
        _backend["name"] = change["new"]
        _roster["instance"] = None
        log(f"Switched backend to {change['new']!r}. Roster reset (empty).")
        refresh_internals()


def on_enroll():
    sid, name, gpa = id_box.value, name_box.value.strip(), gpa_box.value
    get_roster()[sid] = {"name": name, "gpa": gpa}
    log(f"Enrolled/updated {sid}: {get_roster()[sid]}")


def on_lookup():
    sid = id_box.value
    log(f"{sid} -> {get_roster()[sid]}")


def on_drop():
    sid = id_box.value
    del get_roster()[sid]
    log(f"Dropped {sid}.")


def on_list():
    r = get_roster()
    if r.is_empty():
        log("Roster is empty.")
        return
    log(f"{len(r)} student(s) enrolled:")
    for sid in r:
        log(f"  {sid}: {r[sid]}")


def on_load_sample():
    loaded = 0
    for sid, name, gpa in SAMPLE_STUDENTS:
        get_roster()[sid] = {"name": name, "gpa": gpa}
        loaded += 1
    log(f"Loaded {loaded}/{len(SAMPLE_STUDENTS)} sample students.")


def on_reset():
    _roster["instance"] = current_class()()
    log(f"Roster reset (fresh empty {current_class().__name__}).")


backend_toggle.observe(on_backend_change, names="value")
enroll_btn.on_click(guarded(on_enroll))
lookup_btn.on_click(guarded(on_lookup))
drop_btn.on_click(guarded(on_drop))
list_btn.on_click(guarded(on_list))
load_btn.on_click(guarded(on_load_sample))
reset_btn.on_click(guarded(on_reset))
clear_log_btn.on_click(lambda _btn: out.clear_output())

ui = widgets.VBox([
    TITLE_BAR,
    widgets.HBox([backend_toggle], layout=widgets.Layout(margin="14px 0 0 0")),
    widgets.HBox([id_box, name_box, gpa_box], layout=widgets.Layout(margin="10px 0 0 0")),
    widgets.HBox([enroll_btn, lookup_btn, drop_btn, list_btn], layout=widgets.Layout(margin="8px 0 0 0")),
    widgets.HBox([load_btn, reset_btn, clear_log_btn], layout=widgets.Layout(margin="6px 0 0 0")),
    section_header("Action log"),
    out,
    section_header("Hash table internals"),
    internals_out,
    FOOTER,
], layout=widgets.Layout(
    width="900px",
    border_top=BORDER, border_right=BORDER, border_bottom=BORDER, border_left=BORDER,
    padding="0 16px 16px 16px",
))
refresh_internals()
display(ui)
````
